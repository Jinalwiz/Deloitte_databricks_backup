```sql
SELECT DISTINCT
  `App Name`,
  `App ID`,
  CONCAT(
    'https://fioriappslibrary.hana.ondemand.com/sap/fix/externalViewer/#/detail/Apps(',
    "'",
    `App ID`,
    "'",
    ')'
  ) AS `Hyperlink`,
  `UI Technology`,
  `App Description`
FROM
  `jeffdemo`.`harish_demo`.`v_fiori_app_list`
WHERE
  (
    `App Description` ILIKE '%balance sheet%'
    OR `App Name` ILIKE '%balance sheet%'
    OR `App Description` ILIKE '%reconciliation%'
    OR `App Name` ILIKE '%reconciliation%'
    OR `App Description` ILIKE '%financial statement%'
    OR `App Name` ILIKE '%financial statement%'
    OR `App Description` ILIKE '%validation%'
    OR `App Name` ILIKE '%validation%'
  )
  AND `App Name` IS NOT NULL
  AND `App ID` IS NOT NULL
  AND `App Description` IS NOT NULL
  AND `UI Technology` IS NOT NULL
```
