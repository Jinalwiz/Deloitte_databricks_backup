```sql
WITH ranked_customers AS (
  SELECT
    `CustomerText` AS customer_name,
    SUM(`BillingQuantity`) AS order_volume,
    `OverallBillingStatus` AS risk_indicator,
    RANK() OVER (ORDER BY SUM(`BillingQuantity`) DESC) AS rnk
  FROM
    `jeffdemo`.`dlife`.`v_billing_document_item_01`
  WHERE
    `SalesOrganizationName` = 'Pharma US West'
    AND YEAR(`BillingDate`) = 2025
    AND `CustomerText` IS NOT NULL
  GROUP BY
    `CustomerText`,
    `OverallBillingStatus`
)
SELECT
  customer_name,
  order_volume,
  risk_indicator
FROM
  ranked_customers
WHERE
  rnk <= 5
ORDER BY
  order_volume DESC
```
