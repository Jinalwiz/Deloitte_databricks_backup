```sql
SELECT
    DATE_TRUNC('MONTH', 'BillingDate') AS MonthStart,
    SUM(COALESCE('Billed_Revenue', 0)) AS Total_Sales_Revenue,
    SUM(COALESCE('BillingQuantity', 0)) AS Total_Sales_Quantity
  FROM
    'jeffdemo'.'dlife'.'v_billing_document_item_01'
  WHERE
    'BillingDate' >= '2024-01-01'
    AND 'BillingDate' < '2025-01-01'
  GROUP BY
    1
```
