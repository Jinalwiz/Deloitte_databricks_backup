```sql
SELECT
  h.`Level1Name`,
  h.`Level2Name`,
  h.`Level3Name`,
  h.`Level4Name`,
  h.`Level5Name`,
  h.`Level6Name`,
  h.`Level7Name`,
  h.`Level8Name`,
  h.`Level9Name`,
  h.`Level10Name`,
  h.`GLAccount`,
  c.`GLAccountName`,
  SUM(c.`AmountInCompanyCodeCurrency`) AS TotalAmountInCompanyCodeCurrency
FROM
  `jeffdemo`.`dlife`.`v_gl_cube` c
    JOIN `jeffdemo`.`dlife`.`v_gl_hierarchy` h
      ON c.`GLAccount` = h.`GLAccount`
WHERE
  h.`GLAccountHierarchyName` ILIKE '%PHC1%'
GROUP BY
  h.`Level1Name`,
  h.`Level2Name`,
  h.`Level3Name`,
  h.`Level4Name`,
  h.`Level5Name`,
  h.`Level6Name`,
  h.`Level7Name`,
  h.`Level8Name`,
  h.`Level9Name`,
  h.`Level10Name`,
  h.`GLAccount`,
  c.`GLAccountName`
ORDER BY
  h.`Level1Name`,
  h.`Level2Name`,
  h.`Level3Name`,
  h.`Level4Name`,
  h.`Level5Name`,
  h.`Level6Name`,
  h.`Level7Name`,
  h.`Level8Name`,
  h.`Level9Name`,
  h.`Level10Name`,
  h.`GLAccount`;
```
