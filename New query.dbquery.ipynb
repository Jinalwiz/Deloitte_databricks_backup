```sql
SELECT
  c.`GLAccount`,
  h.`Level1Name`,
  h.`Level2Name`,
  h.`Level3Name`,
  h.`Level4Name`,
  h.`Level5Name`,
  h.`Level6Name`,
  h.`Level7Name`,
  h.`Level8Name`,
  h.`Level9Name`,
  h.`Level10Name`,
  SUM(c.`AmountInCompanyCodeCurrency`) AS TotalAmountInCompanyCodeCurrency
FROM
  `jeffdemo`.`dlife`.`v_gl_cube` c
    LEFT OUTER JOIN `jeffdemo`.`dlife`.`v_gl_hierarchy` h
      ON c.`GLAccount` = h.`GLAccount`
WHERE
  c.`CompanyCode` = 'PHC1'
  AND c.`FiscalYear` = '2025'
  AND h.`GLAccountHierarchyName` ILIKE '%PHC1%'
group BY
  c.`GLAccount`,
  h.`Level1Name`,
  h.`Level2Name`,
  h.`Level3Name`,
  h.`Level4Name`,
  h.`Level5Name`,
  h.`Level6Name`,
  h.`Level7Name`,
  h.`Level8Name`,
  h.`Level9Name`,
  h.`Level10Name`;
```
