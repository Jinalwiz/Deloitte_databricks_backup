```python
import json
import requests
from typing import Dict, Any, List

# -----------------------------
# CONFIG
# -----------------------------
OPENAI_API_KEY = "YOUR_API_KEY"
GPT_MODEL = "gpt-5"

SAP_RPT_TRAIN_API = "https://sap-rpt/train"
SAP_RPT_PREDICT_API = "https://sap-rpt/predict"


# -----------------------------
# GPT-5 CALL
# -----------------------------
def call_gpt(prompt: str) -> Dict[str, Any]:
    headers = {
        "Authorization": f"Bearer {OPENAI_API_KEY}",
        "Content-Type": "application/json"
    }

    payload = {
        "model": GPT_MODEL,
        "messages": [
            {"role": "system", "content": "You are a data engineer converting analytics output into relational ML datasets."},
            {"role": "user", "content": prompt}
        ],
        "response_format": {"type": "json_object"}
    }

    response = requests.post(
        "https://api.openai.com/v1/chat/completions",
        headers=headers,
        json=payload
    )

    return json.loads(response.json()["choices"][0]["message"]["content"])


# -----------------------------
# TRANSFORM AIBI OUTPUT
# -----------------------------
def transform_aibi_output(aibi_output: Dict[str, Any],
                          prediction_context: Dict[str, Any]) -> Dict[str, Any]:
    """
    Converts ANY AIBI Genie output into:
    - Training dataset
    - Prediction dataset
    """

    prompt = f"""
Given the following analytics output and prediction request,
convert it into RELATIONAL ML DATA suitable for SAP RPT models.

Rules:
1. Identify entities, time dimensions, measures automatically.
2. Output must be schema-agnostic.
3. Training data must contain historical rows.
4. Prediction data must contain future rows with missing target.
5. Do NOT assume specific KPIs.

Analytics Output:
{json.dumps(aibi_output, indent=2)}

Prediction Request:
{json.dumps(prediction_context, indent=2)}

Return JSON with:
- entities
- features
- target
- training_data (list of rows)
- prediction_data (list of rows)
"""

    return call_gpt(prompt)


# -----------------------------
# TRAIN RPT MODEL
# -----------------------------
def train_rpt_model(dataset: Dict[str, Any]) -> str:
    payload = {
        "entities": dataset["entities"],
        "features": dataset["features"],
        "target": dataset["target"],
        "data": dataset["training_data"]
    }

    response = requests.post(SAP_RPT_TRAIN_API, json=payload)
    return response.json().get("model_id")


# -----------------------------
# PREDICT USING RPT MODEL
# -----------------------------
def predict_rpt(model_id: str, dataset: Dict[str, Any]) -> Dict[str, Any]:
    payload = {
        "model_id": model_id,
        "data": dataset["prediction_data"]
    }

    response = requests.post(SAP_RPT_PREDICT_API, json=payload)
    return response.json()


# -----------------------------
# END-TO-END PIPELINE
# -----------------------------
def run_pipeline(aibi_output: Dict[str, Any],
                 prediction_context: Dict[str, Any]) -> Dict[str, Any]:

    transformed = transform_aibi_output(aibi_output, prediction_context)

    model_id = train_rpt_model(transformed)

    predictions = predict_rpt(model_id, transformed)

    return {
        "model_id": model_id,
        "predictions": predictions
    }
```
